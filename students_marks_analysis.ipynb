```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Students Marks Analysis Project Started!")
```

```python

data = {
    "Student": ["Ali", "Sara", "Ahmed", "Ayesha", "Usman",
                "Fatima", "Hassan", "Zainab", "Bilal", "Maryam"],
    
    "Math": [85, 92, 67, 95, 72, 88, 76, 90, 65, 98],
    
    "English": [78, 89, 75, 91, 68, 84, 80, 93, 70, 96],
    
    "Science": [90, 94, 72, 96, 75, 91, 82, 88, 69, 97]
}

df = pd.DataFrame(data)

df
```

```python

df["Total"] = df[["Math", "English", "Science"]].sum(axis=1)

df
```

```python

df["Percentage"] = (df["Total"] / 300) * 100

df
```

```python

df["Percentage"] = df["Percentage"].round(2)

df
```

```python

average_marks = df["Total"].mean()

print("Class Average Marks:", round(average_marks, 2))
```

```python

top_student = df.loc[df["Total"].idxmax()]

print("Top Student:", top_student["Student"])
print("Total Marks:", top_student["Total"])
print("Percentage:", top_student["Percentage"])
```

```python

df["Result"] = np.where(
    df["Percentage"] >= 50,
    "Pass",
    "Fail"
)

df[["Student", "Total", "Percentage", "Result"]]
```

```python

plt.figure(figsize=(12, 6))

plt.bar(df["Student"], df["Total"])

plt.title("Students Total Marks Analysis")
plt.xlabel("Student Name")
plt.ylabel("Total Marks out of 300")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python

subjects = ["Math", "English", "Science"]

average_scores = [
    df["Math"].mean(),
    df["English"].mean(),
    df["Science"].mean()
]

for subject, average in zip(subjects, average_scores):
    print(f"{subject} Average: {average:.2f}")
```

```python

plt.figure(figsize=(8, 5))

plt.bar(subjects, average_scores)

plt.title("Average Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks out of 100")

plt.tight_layout()
plt.show()
```

```python

df.to_csv("students_marks_results.csv", index=False)

print("Student results saved successfully!")
```
